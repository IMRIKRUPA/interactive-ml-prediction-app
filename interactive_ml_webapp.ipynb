## 1. Application Architecture & Verification
Verifying that the trained model pipeline loads seamlessly and performs inference on arbitrary inputs.


````
import os
import joblib
import pandas as pd
import numpy as np

# Load the trained model pipeline
model_path = "models/titanic_classifier.joblib"
model = joblib.load(model_path)
print("Pipeline loaded successfully:", type(model))
````

````
# Testing inference with sample passenger profiles
sample_passengers = pd.DataFrame([
    {"pclass": 1, "sex": "female", "age": 29.0, "fare": 100.0, "embarked": "S", "family_size": 2},
    {"pclass": 3, "sex": "male", "age": 22.0, "fare": 7.89, "embarked": "S", "family_size": 1}
])

preds = model.predict(sample_passengers)
probas = model.predict_proba(sample_passengers)

for idx, row in sample_passengers.iterrows():
    status = "SURVIVED" if preds[idx] == 1 else "DID NOT SURVIVE"
    print(f"Passenger {idx+1} ({row['sex']}, Class {row['pclass']}): Prediction = {status} | Survival Probability = {probas[idx][1]*100:.1f}%")
````
